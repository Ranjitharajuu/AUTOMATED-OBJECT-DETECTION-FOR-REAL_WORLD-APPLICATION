# Data Loading
Load the MNIST dataset using Keras.

```python
# import Necessary Libraries
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras import layers
```

```python
# Load dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Display dataset shapes
print(f"x_train shape: {x_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"x_test shape: {x_test.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
x_train[0]
```

```python
# Basic data visualisation
import matplotlib.pyplot as plt
import numpy as np
plt.figure(figsize=(10, 4))
for i in range(10):
  plt.subplot(2, 5, i + 1)
  plt.imshow(x_train[i], cmap='gray')
  plt.xticks([])
  plt.yticks([])
  plt.title(y_train[i])
plt.tight_layout()
```

# Data Preprocessing
Normalize the data and reshape it for CNN input.

```python
# Normalize and reshape
x_train_nom = x_train/255
x_test_nom= x_test/255
```

```python
len(np.unique(y_train))
```

```python
from tensorflow.keras.utils import to_categorical
y_train_cat=to_categorical(y_train,num_classes=len(np.unique(y_train)))
y_test_cat=to_categorical(y_train,num_classes=len(np.unique(y_train)))
y_train_cat.shape,y_test_cat.shape
```

```python
x_train_nom = x_train_nom.reshape(-1,28,28,1)
x_train_nom[0]
```

# Model Building
Build a CNN model using Keras Sequential API.

```python
model = Sequential()
model.add(layers.Conv2D(16,3, padding='same',activation='relu',input_shape=(28,28,1)))
model.add(layers.MaxPool2D(2))
model.add(layers.Conv2D(32,3, padding='same',activation='relu',input_shape=(28,28,1)))
model.add(layers.MaxPool2D(2))
model.add(layers.Conv2D(64,3, padding='same',activation='relu',input_shape=(28,28,1)))
model.add(layers.MaxPool2D(2))
model.add(layers.Flatten())
model.add(layers.Dense(128,activation='relu'))
model.add(layers.Dense(10,activation='softmax'))
model.summary()
```

# Model Compilation
Compile the model with Adam optimizer and categorical crossentropy loss.

```python
model.compile(optimizer="adam", loss='categorical_crossentropy', metrics=['accuracy'])
```

# Model Training
Train the model on the training data.

```python
history = model.fit(x_train_nom, y_train_cat, epochs=5, batch_size=128, validation_split=0.2)
```

# Model Evaluation
Evaluate the model on the test data.

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)
print(f"Test Loss: {test_loss}")
print(f"Test Accuracy: {test_accuracy}")
```

# Prediction
Make predictions on new data samples.

```python
# Predict on a few test samples
predictions = model.predict(x_test_nom[:10])
predicted_classes = np.argmax(predictions, axis=1)
true_classes = np.argmax(y_test_cat[:10], axis=1)

# Display predictions
plt.figure(figsize=(10, 2))
for wind, i in enumerate(range(10)):
    plt.subplot(1, 10, wind+1)
    plt.imshow(x_test[i].reshape(28, 28), cmap='gray')
    plt.xlabel(f"True: {true_classes[i]}, Predicted: {predicted_classes[i]}",
              color='g' if predicted_classes[i]!=true_classes[i] else 'g',
              rotation=45)
    plt.xticks([])
    plt.yticks([])
plt.tight_layout()
```
